```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
# Load dataset from root folder (has upazila column)
df = pd.read_csv("../smartgrid_risk_dataset.csv")
print(df.shape)
df.head()
```

```python
print(df.isnull().sum())
```

```python
print(df["risk_level"].value_counts())
print(df["risk_level"].value_counts(normalize=True))
```

```python
# Drop datetime, risk_score, and date features (year, month, day)
df = df.drop(columns=[
    "datetime",
    "risk_score",
    "year",
    "month",
    "day"
])

print("After dropping columns:", df.shape)
```

```python
categorical_cols = [
    "weather_state",
    "district",
    "upazila",
    "area_type",
    "substation_id",
    "feeder_id",
    "maintenance_due"
]

encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le
```

```python
target_encoder = LabelEncoder()
df["risk_level"] = target_encoder.fit_transform(df["risk_level"])
```

```python
X = df.drop("risk_level", axis=1)
y = df["risk_level"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
import numpy as np
import os

os.makedirs("../models", exist_ok=True)

# Save unscaled data for Decision Tree and Random Forest
np.save("../models/X_train.npy", X_train)
np.save("../models/X_test.npy", X_test)
np.save("../models/y_train.npy", y_train)
np.save("../models/y_test.npy", y_test)

print("Train/Test datasets saved successfully.")
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training Shape :", X_train_scaled.shape)
print("Testing Shape  :", X_test_scaled.shape)
```

```python
import os
import joblib

os.makedirs("../models", exist_ok=True)

joblib.dump(scaler, "../models/scaler.pkl")
joblib.dump(target_encoder, "../models/target_encoder.pkl")

print("Saved successfully!")
```
